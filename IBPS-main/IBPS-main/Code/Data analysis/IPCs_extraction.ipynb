```python
import re

def extract_ipc_numbers(user_text):
    # Match numbers with optional uppercase letter suffix (e.g., 29A, 120B)
    matches = re.findall(r'\b\d+[A-Z]?\b', user_text.upper())
    return matches
```

```python
def get_ipc_descriptions(section_ids, ipcs_data):
    descriptions = {}
    for section in section_ids:
        ipc = ipcs_data.get(section)
        if ipc:
            # Use section number as key; can also include title
            descriptions[section] = ipc['description']
    return descriptions
```

```python
relevant_descriptions = get_ipc_descriptions(section_ids, ipcs_data)
```

```python
for section, desc in relevant_descriptions.items():
    print(f"Section {section}: {desc}\n")
```

```python
def extract_and_get_ipc_context(user_input, ipcs_data):
    section_ids = extract_ipc_numbers(user_input)
    return get_ipc_descriptions(section_ids, ipcs_data)

# Example
user_input = " [\\\"Section 438 CrPC\\\", \\\"Section 376(2)(N) IPC\\\", \\\"Section 4 of Prevention of Children from Sexual Offences, 2012\\\"]"
context = extract_and_get_ipc_context(user_input, ipcs_data)
for sec, desc in context.items():
    print(f"Section {sec}: {desc}")
```

```python
import re
import json
import pandas as pd
from tqdm import tqdm

# Sample loading logic
# with open('your_json_file.json', 'r', encoding='utf-8') as f:
#     data = json.load(f)

def extract_case_info(entry):
    case_text = entry['case']

    # 1. Bail type
    bail_type_match = re.search(r'Applicant applied for ([\w-]+)', case_text)
    bail_type = bail_type_match.group(1) if bail_type_match else ""

    # 11.Case details
    details_match = re.search(r'Details of the incident are (.*)', case_text, re.DOTALL)
    details_of_incident = details_match.group(1).strip() if details_match else "None"

    # 2. Withdrawal application
    withdrawal_match = re.search(r'Is it a withdrawal application\?\s*(Yes|No)', case_text, re.IGNORECASE)
    is_withdrawal = withdrawal_match.group(1).capitalize() if withdrawal_match else "Unknown"

    # 3. Age
    age_match = re.search(r'Age of the accused is\s*(\d+)', case_text)
    age = age_match.group(1) if age_match else "Unknown"

    # 4. Health issues
    health_match = re.search(r'Health issues for the accused are\s*([^\n.]+)', case_text)
    health_issues = health_match.group(1).strip() if health_match else "Unknown"

    # 5. Past criminal records
    if re.search(r'There are no past criminal records of the accused', case_text, re.IGNORECASE):
        past_record = "No"
    elif re.search(r'There are (some\s+)?past criminal records of the accused', case_text, re.IGNORECASE):
        past_record = "Yes"
    else:
        past_record = "Unknown"

    # 6. Statutes
    # statutes_match = re.search(r'Statutes mentioned in the judgement are\s*\[([^\]]+)\]', case_text)
    # statutes = [s.strip() for s in statutes_match.group(1).split(',')] if statutes_match else []
    match_text = re.search(r'Statutes mentioned in the judgement are\s*\[(.*?)\]', case_text)
    if match_text:
        content = match_text.group(1).strip()
        if not content:
            statutes= []  # Empty brackets
        else:
            statutes= [s.strip() for s in content.split(',')]
    else:
        statutes='no_match'

    # 7. Outcome
    outcome_ex = entry.get('outcome', '').lower()
    outcome_text=re.findall(r'[^.]+', outcome_ex)
    if(outcome_text):
        if "bail granted" in outcome_text[0]:
            outcome = "Granted"
        elif "bail not granted" in outcome_text[0]:
            outcome = "Not Granted"
        elif "bail cancelled" in outcome_text[0]:
            outcome = "Cancelled"
        elif "bail not cancelled" in outcome_text[0]:
            outcome = "Not Cancelled"
        else:
            outcome = "Unknown"
    else :
        outcome = "Unknown"

    # 8. Date of Arrest
    date_of_arrest = entry.get('date_of_arrest', "Unknown")

    # 9. Date of Judgement
    date_of_judgement = entry.get('date_of_judgement', "Unknown")

    #10. CNR number extraction
    cnr_num=entry.get('CNR', "Unknown")

    return {
        'CNR': cnr_num,
        'bail_type': bail_type,
        'is_withdrawal': is_withdrawal,
        'age': age,
        'health_issues': health_issues,
        'past_criminal_record': past_record,
        'statutes': statutes,
        'outcome': outcome,
        'date_of_arrest': date_of_arrest,
        'date_of_judgement': date_of_judgement,
        'case_details_length': len(details_of_incident.split())
        # 'case_detals': details_of_incident
    }



# Open and read the JSON file
with open('/content/drive/MyDrive/IBPS/final_cleaned_alt[2].json', 'r') as file:
    data = json.load(file)

data2=data[0 : 5]
# Parse all cases
df_data = [extract_case_info(entry) for entry in tqdm(data,desc="Processing Cases")]

# Create DataFrame
df = pd.DataFrame(df_data)

# Show or save
df.head()
df.to_csv("extracted_case_info_with_case_len2.csv", index=False)
```
