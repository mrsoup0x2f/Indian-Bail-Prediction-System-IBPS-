```python
import os
import json
import pdfplumber
import fitz
# Folder containing PDF files
PDF_FOLDER = 'PDF_Rajasthan'  # Change this to your folder path
OUTPUT_JSON = 'rajasthan_raw_json.json'
BATCH_SIZE = 128

# def extract_text_from_pdf(file_path):
#     try:
#         with pdfplumber.open(file_path) as pdf:
#             return "\n".join(page.extract_text() or "" for page in pdf.pages)
#     except Exception as e:
#         print(f"Error reading {file_path}: {e}")
#         return ""
def extract_text_from_pdf(file_path):
    try:
        with fitz.open(file_path) as doc:
            return "\n".join(page.get_text() for page in doc)
    except Exception as e:
        print(f"Error reading {file_path}: {e}")
        return ""
def process_pdfs_in_batches(folder_path, batch_size, output_file):
    all_files = [f for f in os.listdir(folder_path) if f.lower().endswith('.pdf')]
    batch = []
    total_written = 0

    # Initialize or clear the output file
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write('[')

    for idx, file_name in enumerate(all_files):
        file_path = os.path.join(folder_path, file_name)
        text = extract_text_from_pdf(file_path)

        batch.append({
            "CNR": os.path.splitext(file_name)[0],
            "text": text.strip()
        })

        # Write batch if full or last file
        is_last_file = (idx == len(all_files) - 1)
        if len(batch) == batch_size or is_last_file:
            with open(output_file, 'a', encoding='utf-8') as f:
                for i, item in enumerate(batch):
                    if total_written > 0 or i > 0:
                        f.write(",\n")
                    json.dump(item, f, ensure_ascii=False)
            total_written += len(batch)
            print(f"Wrote {len(batch)} files to JSON (Total: {total_written})")
            batch = []

    # Close the JSON array
    with open(output_file, 'a', encoding='utf-8') as f:
        f.write(']')

if __name__ == '__main__':
    process_pdfs_in_batches(PDF_FOLDER, BATCH_SIZE, OUTPUT_JSON)
```
